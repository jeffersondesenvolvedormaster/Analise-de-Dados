```python
import math
n = float(input('Informe um número: '))
x = n * math.pi
print('x = n * pi = ', x)
print('Teto de x =', math.ceil(x))
print('Piso de x =', math.floor(x))
print('Log de x na base 10 =', math.log(x, 10))
print('Raiz de x =', math.sqrt(x))
```
