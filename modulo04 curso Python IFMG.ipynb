cria uma lista usando numeros digitados entre espaço durante um input

```python


def criarLista():
    lista = []
    while True:
        entrada = input("Digite um número inteiro para incluir na lista ou s para finalizar: ")
        if entrada == 's':
            break
        elif entrada:
            x=float(entrada)
            lista.append(x)
    return lista       
    
criarLista()
```

A cada input adiciona um numero a lista 
Mais pratico pois digita o numero e aperta enter e só termina quando digita s para sair (break)

```python
criarLista()
```

```python
precoProdutos= criarLista()
print('preço dos produtos = R$',precoProdutos)
```

```python
fornecedores=2

for cont in range(fornecedores):
    print(f'Fornecedor {cont + 1}:')
    precoProdutos= criarLista()
    print('Preços dos produtos são: R$', precoProdutos)
    print('Preço médio dos produtos é: R$', sum(precoProdutos) / len(precoProdutos))
    print('o valor total dos produtos é: R$', sum(precoProdutos))
    print('\n')
```

```python
lista_num = []

while True:
    n = int(input('Informe um número (0 para parar):'))
    if n == 0:
        break
    lista_num.append(n) # coloca um numero na lista lista_num
    
dic_num = {}

for n in lista_num: #pelo numero de itens na list_num
    if n in dic_num: #se tiver o mesmo numero do input dentro do dic_num
        dic_num[n] += 1 #acressente o input n +1 a dic_num (quantidade de vezes que aparece o numero)
        
    else:
        dic_num[n] = 1 #quando o imput for zero ele para o loop e faz o que esta abaixo:

for n, quant in dic_num.items():
    print('Número', n, 'aparece:', quant,'vezes')
    
print(dic_num)
print(lista_num)
```
