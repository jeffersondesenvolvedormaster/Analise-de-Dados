Ao usar um loop: enquanto ocorrer um erro o except é acionado e o loop retorna ao inicio, caso o loop seja finalizado corretamente da prosseguimento ao código.

```python
while True:
    try:
        print('Informe dois números')
        n1 = float(input('n1: '))
        n2 = float(input('n2: '))
        r = n1 / n2
        break
    except Exception as e:
        print('Ocorreu o seguinte erro:', type(e))
print(n1, '/', n2, '=', r)
```

A clausula except Exception é nomeada como 'e' 
caso haja o erro acontece um print que informa o tipo do erro com a função type que recebe como parametro 'e'
Dessa forma podemos mostrar uma mensagem com o tipo de erro e até mesmo criar uma resposta para cada tipo de erro.

```python
while True:
    try:
        print('Informe dois números')
        n1 = float(input('n1: '))
        n2 = float(input('n2: '))
        r = n1 / n2
        break
    except ValueError:
        print('Número inválidos! Tente novamente.')
    except ZeroDivisionError as e:
        print(e)
        print('Divisão por zero! Tente novamente.')
print(n1, '/', n2, '=', r)
```

Podemos criar uma except para cada tipo de erro 
podemos criar um alias (tipo de apelido exemplo ValueError = e print(e)) para o except e mandar imprimir o erro 
Ou ao ocorrer um erro executar uma ação como um print ou qualquer outra ação é disparada a ocorrer aquele tipo de erro

```python
try:
    # Código a ser tentado
    numero1 = int(input("Digite o primeiro número: "))
    numero2 = int(input("Digite o segundo número: "))
    resultado = numero1 / numero2
    print('o resultado é' ,resultado)
except ZeroDivisionError:
    # Bloco executado apenas se ocorrer ZeroDivisionError
    print("Erro: Divisão por zero!")
else:
    # Bloco executado APENAS se NENHUM erro ocorrer no bloco try
    print("A operação de divisão foi um sucesso.")
finally:
    # Bloco executado SEMPRE, independentemente de erro
    print("Fim da execução do bloco try-except-else-finally.")
```

A instrução try possui também as cláusulas else e finally. A cláusula else pode ser usada
para realizar alguma ação quando não ocorrer erros. Já a cláusula finally é executa
incondicionalmente, ocorrendo erros ou não. A cláusula finally é útil para executar ações de limpeza
como fechamento de arquivos.

A função juros recebe três parametros um deles com valor defaut,
retorna uma operação matematica com os parametros,

passa por uma condicional para verificar se o parametro que recebeu valor defaut foi modificada ou não

os parametros são alterados por variaveis inputadas,

```python
def fat(num):
    if num <= 1:
        return 1
    else:
        return num * fat(num-1)
n = int(input('Informe um número: '))
print('O fatorial de', n, 'é', fat(n))
```

Acima é um exemplo de  recursão que ocorre quando uma função chama a si mesma. Na prática, é possível usar laços

com a condição de parada, assim como fazemos nos laços de repetição. Caso contrário, podemos
cair em uma recursão infinita que a função faz chamadas a ela mesma indefinidamente.

```python
def cubo(num):
    return num * num * num

def fat(num):
    if num <= 1:
        return 1
    return num * fat(num - 1)
```

a função calcula retorna uma eval (calcula uma expressão como se fosse uma string usando aspas)

```python


HIST = '' # define a variavel global vazia (vai receber dados posteriormente)

def calcula(expr): # função que calcula a expressão
    try: # tenta executar o código
        return eval(expr)
    except: #
        print('Expressão inválida!')
        return None
    
def historico(expr, res): # função que armazena o histórico
    global HIST # indica que a variável HIST é global e acessa a variável fora da função
    if res is not None: # verifica se o resultado não é None
        HIST += '\n\n'+ expr # adiciona a expressão ao histórico
        HIST += '\n' + str(res) # adiciona o resultado ao histórico
    
def principal():
    while True:
        expr = input('Digite uma expressão: ou h para histórico, s para sair') # parametro da função calcula
        if expr.lower() == 's': # sair do programa
            break
        if expr.lower() == 'h':# acessar o histórico
            print(HIST, '\n')
        else:
            res = calcula(expr) # calcula a expressão
            historico(expr, res) # armazena no histórico
            print('o resultado da expressão é: ',res, '\n')
            
if __name__ == '__main__': # ponto de entrada do programa, verifica se o script está sendo executado diretamente
    principal() # inicia a função principal
```
