```python
soma = 0
while True:
    n = float(input('Informe um número: '))
    if n < 0:
        continue
    if n == 0:
        break
    soma = soma + n
    print('Soma dos números:', soma)
```

```python
print('Informe dois números')
n1 = int(input('N1: '))
n2 = int(input('N2: '))

if n1 < 1 or n2 < 1:
    print('Números inválidos para MDC')
else:
    while True:
        resto = n1 % n2
        print (n1, '/' , n2, '-> resto:', resto)
        if resto == 0:
            break
        n1 = n2
        n2 = resto
    print('O MDC é ', n2)
```

resto é igual a sobra da divisão de N1 e N2
a cada volta no laço imprime N1 e N2 e a variavel resto
a cada volta no laço N1 vira N2 e N2 vira o resto
a cada volta no laço os novos parametros N1 e N2 são divididos e a sobra vira o nvo valor da variavel 'resto'

```python
n = int(input('Informe o número de elementos do conjunto: '))

print('Elementos:', end=' ')
for cont in range(1, n+1):
    print(cont, end=' ')
    
print('\nSubconjuntos:', end='')
for cont in range(1, n+1):
    for cont2 in range(cont+1, n+1):
        print('{', cont, ', ', cont2, '}', sep='', end=' ')
```

```python
maior = float('-inf')
menor = float('inf') 
while True:
    n = float(input('Informe um número: '))
    if n > maior:
        maior = n
    if n < menor:
        menor = n
    resp = input('Deseja continuar? (S/N)')
    if resp.lower() == 'n':
        break
print('Maior número informado:', maior)
print('Menor número informado:', menor)
```
