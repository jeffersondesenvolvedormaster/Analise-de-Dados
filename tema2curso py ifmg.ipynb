ABAIXO COMEÇA O ESTUDO DE ESTRUTURAS DE LAÇO (REPETIÇÃO)

```python
atual = 0
n = int(input('Informe um número: '))
while atual < n:
    print(atual)
    atual += 2
```

acima informamos N que while(enquanto) o valor da variavel ATUAL não atingir a condição declarada, aciona um print e acrescenta 2 a variavel, quando atinge a condição termina o loop(laço).

```python
soma = 1
for cont in range(10):
    n = float(input('Informe um número: '))
    soma += n
print(soma)
```

acima a variavel (soma) vai ser acrescida pelo input(n) o numero de vezes informado no loop (for cont in range(10)) que significa faça essa ação na contagen de 10 vezes, no caso essa acão é um pedido de input e a soma do input(n).

```python
soma = {a,b,c}
for r in soma:
    n = int(input('Informe um número: '))
    print(f'valor digitado {n}, soma com indice lista {n+r}')
```

acima podemos notar que o lopp (for) usa o (r) como uma variavel do indice da lista da variavel (soma) evocada após o (in), após inserimos o imput (n) que vai ser somado ao indice da lista(r).

```python
soma = 0
while True:
    n = int(input('Informe um número: '))
    if soma >10:
        break
    soma = soma + n
    print('Soma dos números:', soma)
```

acima usamos um loop onde as condiçoes de inicio e fim são definidas com (true or false) assim se a condição for (true) continua se a condição for false(break)

```python
soma = 0
while True:
    n = float(input('Informe um número: '))
    if n < 0:
        continue
    if n == 0:
        break
    soma = soma + n
    print('Soma dos números:', soma)
```

acima o parametro (continue) desconsidera a condição citada, no caso se o input(n) for menor que 0.
